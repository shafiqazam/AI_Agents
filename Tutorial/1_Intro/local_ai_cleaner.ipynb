```python
import json
import ollama

messy_input = """
User log entry:
ID: #9921
Name: john DOe
Email: jOhn.doe[at]gmail.com
Status: active account since 2024-05-12
Notes: He wants to upgrade to premium but billing failed twice.
"""
```

```python
system_instructions = """
    You are a strict data transformation engine. Your task is to parse unstructured data 
    and return a clean, valid JSON object. Go through each attribute, understand what it means and ensure that the content is properly formatted. 
    For example, if you see an email, make sure it's in the correct format. 
    If you see a date, convert it to ISO format (YYYY-MM-DD). 
    If you see a name, capitalize it properly.
    Do not inculde any conversational text or explanations. Do not add new attributes that are not present in the input.
    Only return raw JSON data. start with  { and end with }.
    """
```

```python
user_input = f"""

Parse the following unstructured data and return a clean JSON object:

Text to parse:
{messy_input}
"""

print("⏳ Sending data to local Llama 3.2...")
```

```python
response = ollama.chat(
    model='llama3.2:3b',
    messages=[
        {'role': 'system', 'content': system_instructions},
        {'role': 'user', 'content': user_input}
    ],
    # options={'temperature': 0.0} # Setting temperature to 0 makes the model highly deterministic
)
```

```python
raw_output = response.message.content.strip()
print("\n🚀 Output from Local LLM:")
print(raw_output)
```

```python
raw_output
```

```python
try:
    structured_json = json.loads(raw_output)
    print("\n✅ Success! The data is now a valid Python dictionary:")
    print(f"User Email: {structured_json['email']}")
except json.JSONDecodeError:
    print("\n❌ Error: The model output wasn't pure JSON. Time to adjust the prompt!")
```
